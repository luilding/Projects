```python
import pandas as pd
import os

def explore_route(project_root):
    data_dir = os.path.join(project_root, 'mpt_data', 'bus')
    stops = pd.read_csv(os.path.join(data_dir, 'stops.txt'))
    stop_times = pd.read_csv(os.path.join(data_dir, 'stop_times.txt'))
    trips = pd.read_csv(os.path.join(data_dir, 'trips.txt'))
    routes = pd.read_csv(os.path.join(data_dir, 'routes.txt'))

    # Get trips between High St and Chapel St stops
    start_stop = 13401  # High St/Punt Rd
    end_stop = 14728    # High St/Chapel St
    
    start_trips = set(stop_times[stop_times['stop_id'] == start_stop]['trip_id'])
    end_trips = set(stop_times[stop_times['stop_id'] == end_stop]['trip_id'])
    common_trips = start_trips & end_trips
    
    if common_trips:
        route_ids = trips[trips['trip_id'].isin(common_trips)]['route_id'].unique()
        bus_routes = routes[routes['route_id'].isin(route_ids)]
        print("\nBus routes between High St/Punt Rd and High St/Chapel St:")
        print(bus_routes[['route_short_name', 'route_long_name']])

current_dir = os.getcwd()
project_root = os.path.dirname(current_dir)
explore_route(project_root)
```

```python
import pandas as pd
import os

def load_gtfs(data_dir):
    stops = pd.read_csv(os.path.join(data_dir, 'stops.txt'))
    routes = pd.read_csv(os.path.join(data_dir, 'routes.txt'))
    trips = pd.read_csv(os.path.join(data_dir, 'trips.txt'))
    stop_times = pd.read_csv(os.path.join(data_dir, 'stop_times.txt'))
    return stops, routes, trips, stop_times

def find_stop(stops_df, location):
    matches = stops_df[stops_df['stop_name'].str.contains(location, case=False, na=False)]
    if matches.empty:
        return None
    return matches.iloc[0]

def find_route(stop_times_df, trips_df, routes_df, start_stop_id, end_stop_id):
    start_trips = set(stop_times_df[stop_times_df['stop_id'] == start_stop_id]['trip_id'])
    end_trips = set(stop_times_df[stop_times_df['stop_id'] == end_stop_id]['trip_id'])
    common_trips = start_trips & end_trips
    
    if not common_trips:
        return None
        
    first_trip = list(common_trips)[0]
    route_id = trips_df[trips_df['trip_id'] == first_trip]['route_id'].iloc[0]
    return routes_df[routes_df['route_id'] == route_id].iloc[0]

def main():
    data_dir = os.path.join(os.path.dirname(os.getcwd()), 'mpt_data', 'bus')
    stops, routes, trips, stop_times = load_gtfs(data_dir)
    
    start = input("Enter start location: ")
    end = input("Enter destination: ")
    
    start_stop = find_stop(stops, start)
    if start_stop is None:
        print(f"No stops found near {start}")
        return
        
    end_stop = find_stop(stops, end)
    if end_stop is None:
        print(f"No stops found near {end}")
        return
        
    print(f"\nStart stop: {start_stop['stop_name']}")
    print(f"End stop: {end_stop['stop_name']}")
    
    route = find_route(stop_times, trips, routes, start_stop['stop_id'], end_stop['stop_id'])
    if route is None:
        print("No direct route found")
    else:
        print(f"\nTake bus {route['route_short_name']}: {route['route_long_name']}")

if __name__ == "__main__":
    main()
```

```python
import pandas as pd
import os

def load_gtfs(data_dir):
    stops = pd.read_csv(os.path.join(data_dir, 'stops.txt'))
    routes = pd.read_csv(os.path.join(data_dir, 'routes.txt'))
    trips = pd.read_csv(os.path.join(data_dir, 'trips.txt'))
    stop_times = pd.read_csv(os.path.join(data_dir, 'stop_times.txt'))
    return stops, routes, trips, stop_times

def find_nearby_stops(stops_df, location, limit=3):
    matches = stops_df[stops_df['stop_name'].str.contains(location, case=False, na=False)]
    # Remove duplicates based on stop_name
    matches = matches.drop_duplicates(subset='stop_name')
    return matches.head(limit)

def find_route_with_transfer(stop_times_df, trips_df, routes_df, stops_df, start_stop_id, end_stop_id):
    # Get all trips for start and end stops
    start_trips = set(stop_times_df[stop_times_df['stop_id'] == start_stop_id]['trip_id'])
    end_trips = set(stop_times_df[stop_times_df['stop_id'] == end_stop_id]['trip_id'])
    
    # Get routes for these trips
    start_route_ids = trips_df[trips_df['trip_id'].isin(start_trips)]['route_id']
    end_route_ids = trips_df[trips_df['trip_id'].isin(end_trips)]['route_id']
    
    start_routes = routes_df[routes_df['route_id'].isin(start_route_ids)]
    end_routes = routes_df[routes_df['route_id'].isin(end_route_ids)]
    
    # Get all stops served by start routes
    start_route_trips = trips_df[trips_df['route_id'].isin(start_routes['route_id'])]
    start_stops = stop_times_df[stop_times_df['trip_id'].isin(start_route_trips['trip_id'])]['stop_id'].unique()
    
    # Get all stops served by end routes
    end_route_trips = trips_df[trips_df['route_id'].isin(end_routes['route_id'])]
    end_stops = stop_times_df[stop_times_df['trip_id'].isin(end_route_trips['trip_id'])]['stop_id'].unique()
    
    # Find common stops for transfer
    transfer_stop_ids = set(start_stops) & set(end_stops)
    transfer_stops = stops_df[stops_df['stop_id'].isin(transfer_stop_ids)]
    
    return (
        [str(x) for x in start_routes['route_short_name'].unique()],
        [str(x) for x in end_routes['route_short_name'].unique()],
        transfer_stops
    )

def main():
    data_dir = os.path.join(os.path.dirname(os.getcwd()), 'mpt_data', 'bus')
    stops, routes, trips, stop_times = load_gtfs(data_dir)
    
    start = input("Enter start location: ")
    end = input("Enter destination: ")
    
    start_stops = find_nearby_stops(stops, start)
    end_stops = find_nearby_stops(stops, end)
    
    if start_stops.empty or end_stops.empty:
        print("No stops found")
        return
    
    print(f"\nStart stops:")
    for _, stop in start_stops.iterrows():
        print(f"- {stop['stop_name']}")
    
    print(f"\nDestination stops:")
    for _, stop in end_stops.iterrows():
        print(f"- {stop['stop_name']}")
    
    for _, start_stop in start_stops.iterrows():
        for _, end_stop in end_stops.iterrows():
            print(f"\nRoute from {start_stop['stop_name']} to {end_stop['stop_name']}:")
            start_routes, end_routes, transfer_stops = find_route_with_transfer(
                stop_times, trips, routes, stops, 
                start_stop['stop_id'], end_stop['stop_id']
            )
            
            print(f"Take bus {' or '.join(start_routes)} from start")
            print(f"Transfer to bus {' or '.join(end_routes)} to destination")
            
            if not transfer_stops.empty:
                print("Possible transfer points:")
                for _, stop in transfer_stops.iterrows():
                    print(f"- {stop['stop_name']}")
            else:
                print("No transfer points found")

if __name__ == "__main__":
    main()
```

```python
import os
import pandas as pd
import networkx as nx
from datetime import datetime
import sqlite3

data_dir = os.path.join(os.path.dirname(os.getcwd()), 'mpt_data', 'bus')

class MPTGTFSProcessor:
    def __init__(self):
        self.db_path = os.path.join(data_dir, 'transit.db')
        self.conn = sqlite3.connect(self.db_path)
        self.graph = nx.MultiDiGraph()
    
    def load_gtfs(self):
        required_files = ['stops.txt', 'routes.txt', 'trips.txt', 
                         'stop_times.txt', 'calendar.txt']
        
        for file in required_files:
            file_path = os.path.join(data_dir, file)
            if not os.path.exists(file_path):
                raise FileNotFoundError(f"Missing required file: {file}")
            
            df = pd.read_csv(file_path)
            table_name = file.replace('.txt', '')
            df.to_sql(table_name, self.conn, if_exists='replace', index=False)
        
        # Create indices
        self.conn.execute('CREATE INDEX IF NOT EXISTS stop_times_trip_id_idx ON stop_times(trip_id)')
        self.conn.execute('CREATE INDEX IF NOT EXISTS stop_times_stop_id_idx ON stop_times(stop_id)')
    
    def build_network(self):
        query = """
        SELECT DISTINCT
            t.route_id,
            st1.stop_id as from_stop,
            st2.stop_id as to_stop,
            st1.departure_time,
            st2.arrival_time,
            r.route_short_name,
            s1.stop_name as from_stop_name,
            s2.stop_name as to_stop_name
        FROM stop_times st1
        JOIN stop_times st2 ON st1.trip_id = st2.trip_id
            AND st1.stop_sequence + 1 = st2.stop_sequence
        JOIN trips t ON st1.trip_id = t.trip_id
        JOIN routes r ON t.route_id = r.route_id
        JOIN stops s1 ON st1.stop_id = s1.stop_id
        JOIN stops s2 ON st2.stop_id = s2.stop_id
        ORDER BY st1.departure_time
        """
        
        edges = pd.read_sql_query(query, self.conn)
        for _, row in edges.iterrows():
            self.graph.add_edge(
                row['from_stop'],
                row['to_stop'],
                route_id=row['route_id'],
                route_name=row['route_short_name'],
                departure=row['departure_time'],
                arrival=row['arrival_time'],
                from_name=row['from_stop_name'],
                to_name=row['to_stop_name']
            )
    
    def find_route(self, from_stop_id, to_stop_id):
        if not nx.has_path(self.graph, from_stop_id, to_stop_id):
            return None
        
        path = nx.shortest_path(self.graph, from_stop_id, to_stop_id)
        route_details = []
        
        for i in range(len(path)-1):
            edges = self.graph.get_edge_data(path[i], path[i+1])
            edge = list(edges.values())[0]  # Get first available route
            route_details.append({
                'route': edge['route_name'],
                'from_stop': edge['from_name'],
                'to_stop': edge['to_name'],
                'departure': edge['departure'],
                'arrival': edge['arrival']
            })
        
        return route_details

    def get_all_stops(self):
        return pd.read_sql_query("SELECT * FROM stops", self.conn)

# Usage example
processor = MPTGTFSProcessor()
processor.load_gtfs()
processor.build_network()

# Find a route between two stops
stops = processor.get_all_stops()
print(f"Available stops: {len(stops)} stops found")

# Example: Get route between first and last stop
if len(stops) >= 2:
    route = processor.find_route(stops.iloc[0]['stop_id'], stops.iloc[-1]['stop_id'])
    if route:
        for step in route:
            print(f"Take route {step['route']} from {step['from_stop']} to {step['to_stop']}")
```

```python
from geopy.geocoders import Nominatim
import pandas as pd
from scipy.spatial import KDTree
from geopy.distance import geodesic
import os

def load_bus_data():
    data_dir = os.path.join(os.path.dirname(os.getcwd()), 'mpt_data', 'bus')
    stops_df = pd.read_csv(os.path.join(data_dir, 'stops.txt'))
    routes_df = pd.read_csv(os.path.join(data_dir, 'routes.txt'))
    trips_df = pd.read_csv(os.path.join(data_dir, 'trips.txt'))
    stop_times_df = pd.read_csv(os.path.join(data_dir, 'stop_times.txt'))
    return stops_df, routes_df, trips_df, stop_times_df

def get_coordinates(address):
    geolocator = Nominatim(user_agent="bus_finder")
    melbourne_bbox = [(-38.5267, 144.5937), (-37.5113, 145.5125)]
    location = geolocator.geocode(f"{address}, Melbourne, Australia", 
                                viewbox=melbourne_bbox, 
                                bounded=True)
    return (location.latitude, location.longitude) if location else None

def find_nearest_stop(lat, lon, stops_df):
    coords = stops_df[["stop_lat", "stop_lon"]].values
    kdtree = KDTree(coords)
    distance, index = kdtree.query([lat, lon])
    nearest_stop = stops_df.iloc[index]
    return nearest_stop

def get_bus_routes(start_stop_id, end_stop_id, stop_times_df, trips_df, routes_df):
    start_trips = set(stop_times_df[stop_times_df['stop_id'] == start_stop_id]['trip_id'])
    end_trips = set(stop_times_df[stop_times_df['stop_id'] == end_stop_id]['trip_id'])
    common_trips = start_trips & end_trips
    
    routes = []
    for trip_id in common_trips:
        trip_stops = stop_times_df[stop_times_df['trip_id'] == trip_id].sort_values('stop_sequence')
        start_seq = trip_stops[trip_stops['stop_id'] == start_stop_id]['stop_sequence'].iloc[0]
        end_seq = trip_stops[trip_stops['stop_id'] == end_stop_id]['stop_sequence'].iloc[0]
        
        if start_seq < end_seq:
            route_id = trips_df[trips_df['trip_id'] == trip_id]['route_id'].iloc[0]
            route_info = routes_df[routes_df['route_id'] == route_id].iloc[0]
            
            routes.append({
                'route': route_info['route_short_name'],
                'name': route_info['route_long_name'],
                'departure': trip_stops[trip_stops['stop_id'] == start_stop_id]['departure_time'].iloc[0],
                'arrival': trip_stops[trip_stops['stop_id'] == end_stop_id]['arrival_time'].iloc[0]
            })
    
    return sorted(routes, key=lambda x: x['departure'])

def find_bus_route(start_location, end_location):
    # Load data
    stops_df, routes_df, trips_df, stop_times_df = load_bus_data()
    
    # Get coordinates
    start_coords = get_coordinates(start_location)
    end_coords = get_coordinates(end_location)
    
    if not start_coords or not end_coords:
        return "Could not find one or both locations"
        
    # Find nearest stops
    start_stop = find_nearest_stop(start_coords[0], start_coords[1], stops_df)
    end_stop = find_nearest_stop(end_coords[0], end_coords[1], stops_df)
    
    # Get available routes
    routes = get_bus_routes(start_stop['stop_id'], end_stop['stop_id'], 
                          stop_times_df, trips_df, routes_df)
    
    if not routes:
        return "No direct routes found"
        
    # Format response
    result = []
    result.append(f"From: {start_stop['stop_name']} To: {end_stop['stop_name']}")
    for route in routes[:3]: # Show top 3 options
        result.append(f"Bus {route['route']} - {route['name']}")
        result.append(f"Departure: {route['departure']} Arrival: {route['arrival']}")
    
    return "\n".join(result)


  

# Example usage
if __name__ == "__main__":
    start = input("Enter starting location: ")
    end = input("Enter destination: ")
    print("\nFinding routes...")
    result = find_bus_route(start, end)
    print(result)
```

```python
def find_routes_with_transfers(start_stop_id, end_stop_id, stop_times_df, trips_df, routes_df, max_transfers=1):
    # Cache trip stops and route info
    trip_stops_cache = {}
    route_info_cache = {}
    
    def get_stops_for_trip(trip_id):
        if trip_id not in trip_stops_cache:
            trip_stops_cache[trip_id] = stop_times_df[stop_times_df['trip_id'] == trip_id].sort_values('stop_sequence')
        return trip_stops_cache[trip_id]

    def get_route_info(trip_id):
        if trip_id not in route_info_cache:
            route_id = trips_df[trips_df['trip_id'] == trip_id]['route_id'].iloc[0]
            route_info_cache[trip_id] = routes_df[routes_df['route_id'] == route_id].iloc[0]
        return route_info_cache[trip_id]

    # Index stop times for faster lookup
    stop_trips_index = stop_times_df.groupby('stop_id')['trip_id'].apply(set).to_dict()
    
    # Find direct routes
    common_trips = stop_trips_index.get(start_stop_id, set()) & stop_trips_index.get(end_stop_id, set())
    
    # Check direct trips first
    for trip_id in common_trips:
        trip_stops = get_stops_for_trip(trip_id)
        start_seq = trip_stops[trip_stops['stop_id'] == start_stop_id]['stop_sequence'].iloc[0]
        end_seq = trip_stops[trip_stops['stop_id'] == end_stop_id]['stop_sequence'].iloc[0]
        
        if start_seq < end_seq:
            route_info = get_route_info(trip_id)
            return [({"route": route_info['route_short_name'], 
                     "departure": trip_stops[trip_stops['stop_id'] == start_stop_id]['departure_time'].iloc[0],
                     "arrival": trip_stops[trip_stops['stop_id'] == end_stop_id]['arrival_time'].iloc[0]}, None)]

    # Find one-transfer routes (only look at next 10 possible transfers)
    results = []
    start_trips = stop_trips_index.get(start_stop_id, set())
    
    for trip_id in list(start_trips)[:10]:
        trip_stops = get_stops_for_trip(trip_id)
        start_seq = trip_stops[trip_stops['stop_id'] == start_stop_id]['stop_sequence'].iloc[0]
        possible_transfers = trip_stops[trip_stops['stop_sequence'] > start_seq]['stop_id'].unique()
        
        for transfer_stop in possible_transfers[:5]:
            transfer_trips = stop_trips_index.get(transfer_stop, set()) & stop_trips_index.get(end_stop_id, set())
            
            for second_trip in transfer_trips:
                second_stops = get_stops_for_trip(second_trip)
                transfer_seq = second_stops[second_stops['stop_id'] == transfer_stop]['stop_sequence'].iloc[0]
                end_seq = second_stops[second_stops['stop_id'] == end_stop_id]['stop_sequence'].iloc[0]
                
                if transfer_seq < end_seq:
                    first_route = get_route_info(trip_id)
                    second_route = get_route_info(second_trip)
                    
                    results.append((
                        {"route": first_route['route_short_name'],
                         "departure": trip_stops[trip_stops['stop_id'] == start_stop_id]['departure_time'].iloc[0]},
                        {"route": second_route['route_short_name'],
                         "departure": second_stops[second_stops['stop_id'] == transfer_stop]['departure_time'].iloc[0],
                         "arrival": second_stops[second_stops['stop_id'] == end_stop_id]['arrival_time'].iloc[0]}
                    ))
                    
                    if len(results) >= 3:
                        return results

    return results
```

```python
from geopy.geocoders import Nominatim
import pandas as pd
from scipy.spatial import KDTree
from geopy.distance import geodesic
import os

def load_bus_data():
    data_dir = os.path.join(os.path.dirname(os.getcwd()), 'mpt_data', 'bus')
    stops_df = pd.read_csv(os.path.join(data_dir, 'stops.txt'))
    routes_df = pd.read_csv(os.path.join(data_dir, 'routes.txt'))
    trips_df = pd.read_csv(os.path.join(data_dir, 'trips.txt'))
    stop_times_df = pd.read_csv(os.path.join(data_dir, 'stop_times.txt'))
    return stops_df, routes_df, trips_df, stop_times_df

def get_coordinates(address):
    geolocator = Nominatim(user_agent="bus_finder")
    melbourne_bbox = [(-38.5267, 144.5937), (-37.5113, 145.5125)]
    location = geolocator.geocode(f"{address}, Melbourne, Australia", 
                                viewbox=melbourne_bbox, 
                                bounded=True)
    return (location.latitude, location.longitude) if location else None

def find_nearest_stop(lat, lon, stops_df):
    coords = stops_df[["stop_lat", "stop_lon"]].values
    kdtree = KDTree(coords)
    distance, index = kdtree.query([lat, lon])
    nearest_stop = stops_df.iloc[index]
    return nearest_stop

def get_bus_routes(start_stop_id, end_stop_id, stop_times_df, trips_df, routes_df):
    start_trips = set(stop_times_df[stop_times_df['stop_id'] == start_stop_id]['trip_id'])
    end_trips = set(stop_times_df[stop_times_df['stop_id'] == end_stop_id]['trip_id'])
    common_trips = start_trips & end_trips
    
    routes = []
    for trip_id in common_trips:
        trip_stops = stop_times_df[stop_times_df['trip_id'] == trip_id].sort_values('stop_sequence')
        start_seq = trip_stops[trip_stops['stop_id'] == start_stop_id]['stop_sequence'].iloc[0]
        end_seq = trip_stops[trip_stops['stop_id'] == end_stop_id]['stop_sequence'].iloc[0]
        
        if start_seq < end_seq:
            route_id = trips_df[trips_df['trip_id'] == trip_id]['route_id'].iloc[0]
            route_info = routes_df[routes_df['route_id'] == route_id].iloc[0]
            
            routes.append({
                'route': route_info['route_short_name'],
                'name': route_info['route_long_name'],
                'departure': trip_stops[trip_stops['stop_id'] == start_stop_id]['departure_time'].iloc[0],
                'arrival': trip_stops[trip_stops['stop_id'] == end_stop_id]['arrival_time'].iloc[0]
            })
    
    return sorted(routes, key=lambda x: x['departure'])

def find_bus_route(start_location, end_location):
    stops_df, routes_df, trips_df, stop_times_df = load_bus_data()
    
    start_coords = get_coordinates(start_location)
    end_coords = get_coordinates(end_location)
    
    if not start_coords or not end_coords:
        return "Could not find one or both locations"
        
    start_stop = find_nearest_stop(start_coords[0], start_coords[1], stops_df)
    end_stop = find_nearest_stop(end_coords[0], end_coords[1], stops_df)
    
    routes = find_routes_with_transfers(start_stop['stop_id'], end_stop['stop_id'], 
                                      stop_times_df, trips_df, routes_df)
    
    if not routes:
        return f"No routes found between {start_stop['stop_name']} and {end_stop['stop_name']}"
        
    result = []
    result.append(f"From: {start_stop['stop_name']} To: {end_stop['stop_name']}\n")
    
    for i, (first_leg, second_leg) in enumerate(routes, 1):
        if second_leg:  # Route with transfer
            result.append(f"Option {i} (with transfer):")
            result.append(f"1. Bus {first_leg['route']} at {first_leg['departure']}")
            result.append(f"2. Transfer to Bus {second_leg['route']} at {second_leg['departure']}")
            result.append(f"Arrival: {second_leg['arrival']}\n")
        else:  # Direct route
            result.append(f"Option {i} (direct):")
            result.append(f"Bus {first_leg['route']}")
            result.append(f"Departure: {first_leg['departure']} Arrival: {first_leg['arrival']}\n")
    
    return "\n".join(result)

def find_routes_with_transfers(start_stop_id, end_stop_id, stop_times_df, trips_df, routes_df, max_transfers=1):
    # Cache trip stops and route info
    trip_stops_cache = {}
    route_info_cache = {}
    
    def get_stops_for_trip(trip_id):
        if trip_id not in trip_stops_cache:
            trip_stops_cache[trip_id] = stop_times_df[stop_times_df['trip_id'] == trip_id].sort_values('stop_sequence')
        return trip_stops_cache[trip_id]

    def get_route_info(trip_id):
        if trip_id not in route_info_cache:
            route_id = trips_df[trips_df['trip_id'] == trip_id]['route_id'].iloc[0]
            route_info_cache[trip_id] = routes_df[routes_df['route_id'] == route_id].iloc[0]
        return route_info_cache[trip_id]

    # Index stop times for faster lookup
    stop_trips_index = stop_times_df.groupby('stop_id')['trip_id'].apply(set).to_dict()
    
    # Find direct routes
    common_trips = stop_trips_index.get(start_stop_id, set()) & stop_trips_index.get(end_stop_id, set())
    
    # Check direct trips first
    for trip_id in common_trips:
        trip_stops = get_stops_for_trip(trip_id)
        start_seq = trip_stops[trip_stops['stop_id'] == start_stop_id]['stop_sequence'].iloc[0]
        end_seq = trip_stops[trip_stops['stop_id'] == end_stop_id]['stop_sequence'].iloc[0]
        
        if start_seq < end_seq:
            route_info = get_route_info(trip_id)
            return [({"route": route_info['route_short_name'], 
                     "departure": trip_stops[trip_stops['stop_id'] == start_stop_id]['departure_time'].iloc[0],
                     "arrival": trip_stops[trip_stops['stop_id'] == end_stop_id]['arrival_time'].iloc[0]}, None)]

    # Find one-transfer routes (only look at next 10 possible transfers)
    results = []
    start_trips = stop_trips_index.get(start_stop_id, set())
    
    for trip_id in list(start_trips)[:10]:
        trip_stops = get_stops_for_trip(trip_id)
        start_seq = trip_stops[trip_stops['stop_id'] == start_stop_id]['stop_sequence'].iloc[0]
        possible_transfers = trip_stops[trip_stops['stop_sequence'] > start_seq]['stop_id'].unique()
        
        for transfer_stop in possible_transfers[:5]:
            transfer_trips = stop_trips_index.get(transfer_stop, set()) & stop_trips_index.get(end_stop_id, set())
            
            for second_trip in transfer_trips:
                second_stops = get_stops_for_trip(second_trip)
                transfer_seq = second_stops[second_stops['stop_id'] == transfer_stop]['stop_sequence'].iloc[0]
                end_seq = second_stops[second_stops['stop_id'] == end_stop_id]['stop_sequence'].iloc[0]
                
                if transfer_seq < end_seq:
                    first_route = get_route_info(trip_id)
                    second_route = get_route_info(second_trip)
                    
                    results.append((
                        {"route": first_route['route_short_name'],
                         "departure": trip_stops[trip_stops['stop_id'] == start_stop_id]['departure_time'].iloc[0]},
                        {"route": second_route['route_short_name'],
                         "departure": second_stops[second_stops['stop_id'] == transfer_stop]['departure_time'].iloc[0],
                         "arrival": second_stops[second_stops['stop_id'] == end_stop_id]['arrival_time'].iloc[0]}
                    ))
                    
                    if len(results) >= 3:
                        return results

    return results 

# Example usage
if __name__ == "__main__":
    start = input("Enter starting location: ")
    end = input("Enter destination: ")
    print("\nFinding routes...")
    result = find_bus_route(start, end)
    print(result)
```
