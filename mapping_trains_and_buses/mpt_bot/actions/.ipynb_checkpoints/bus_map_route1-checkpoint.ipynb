```python
import pandas as pd
import os
from collections import defaultdict, deque

class SimpleTramRouter:
    def __init__(self):
        current_dir = os.path.join('..', 'mpt_data', 'tram')
        
        # Read GTFS files
        self.stops = pd.read_csv(os.path.join(current_dir, 'stops.txt'))
        self.routes = pd.read_csv(os.path.join(current_dir, 'routes.txt'))
        self.trips = pd.read_csv(os.path.join(current_dir, 'trips.txt'))
        self.stop_times = pd.read_csv(os.path.join(current_dir, 'stop_times.txt'))
        
        # Build route network
        self._build_network()
        
    def _build_network(self):
        """Build a network of which stops are connected by which routes"""
        self.stop_routes = defaultdict(set)  # stop_id -> set of route_ids
        self.route_stops = defaultdict(set)  # route_id -> set of stop_ids
        
        # Join stop_times with trips to get route information
        stop_route_data = self.stop_times.merge(
            self.trips[['trip_id', 'route_id']], 
            on='trip_id'
        )
        
        # Build the mappings
        for _, row in stop_route_data.iterrows():
            self.stop_routes[row['stop_id']].add(row['route_id'])
            self.route_stops[row['route_id']].add(row['stop_id'])
    
    def find_nearby_stops(self, location_name, limit=3):
        """Find stops that match the given name"""
        matches = self.stops[self.stops['stop_name'].str.contains(location_name, case=False, na=False)]
        return matches.head(limit)
    
    def find_route(self, start_location, end_location):
        """Find a route between two locations with minimum transfers"""
        # Find matching stops
        start_stops = self.find_nearby_stops(start_location)
        end_stops = self.find_nearby_stops(end_location)
        
        if start_stops.empty or end_stops.empty:
            return "Could not find stops matching the provided locations"
        
        end_stop_ids = set(end_stops['stop_id'])
        
        # Try finding a route from each start stop
        best_route = None
        best_transfers = float('inf')
        
        for _, start_stop in start_stops.iterrows():
            route = self._bfs_route(start_stop['stop_id'], end_stop_ids)
            if route and len(route) - 1 < best_transfers:
                best_route = route
                best_transfers = len(route) - 1
        
        if best_route is None:
            return "No route found between the specified locations"
        
        return self._format_route(best_route)
    
    def _bfs_route(self, start_stop_id, end_stop_ids):
        """Use BFS to find route with minimum transfers"""
        queue = deque([(start_stop_id, [start_stop_id], set())])
        visited = {start_stop_id}
        
        while queue:
            current_stop, path, current_routes = queue.popleft()
            
            # Check if we've reached a destination
            if current_stop in end_stop_ids:
                return path
            
            # Get all routes serving this stop
            routes = self.stop_routes[current_stop]
            
            # For each route from this stop
            for route_id in routes:
                # Get all stops on this route
                for next_stop in self.route_stops[route_id]:
                    if next_stop not in visited:
                        visited.add(next_stop)
                        new_routes = current_routes | {route_id}
                        queue.append((next_stop, path + [next_stop], new_routes))
        
        return None
    
    def _format_route(self, stop_ids):
        """Format the route into human-readable instructions"""
        route_description = []
        current_route = None
        
        for i in range(len(stop_ids) - 1):
            current_stop = stop_ids[i]
            next_stop = stop_ids[i + 1]
            
            # Find route(s) that connect these stops
            current_routes = self.stop_routes[current_stop] & self.stop_routes[next_stop]
            if current_routes:
                route_id = list(current_routes)[0]  # Take the first available route
                route_name = self.routes[self.routes['route_id'] == route_id]['route_short_name'].iloc[0]
                
                if route_name != current_route:
                    current_stop_name = self.stops[self.stops['stop_id'] == current_stop]['stop_name'].iloc[0]
                    if current_route:
                        route_description.append(f"Transfer at {current_stop_name}")
                    route_description.append(f"Take tram {route_name} from {current_stop_name}")
                    current_route = route_name
        
        # Add final destination
        final_stop = self.stops[self.stops['stop_id'] == stop_ids[-1]]['stop_name'].iloc[0]
        route_description.append(f"Arrive at {final_stop}")
        
        return "\n".join(route_description)

# Example usage:
if __name__ == "__main__":
    router = SimpleTramRouter()
    route = router.find_route("Southbank", "Collingwood")
    print(route)
```

```python
import pandas as pd
import os
from collections import defaultdict, deque
import time
import folium
from datetime import datetime
import webbrowser

class SimpleTramRouter:
    def __init__(self):
        start_time = time.time()
        print("Starting router initialization...")
        
        current_dir = os.path.join('..', 'mpt_data', 'tram')
        
        # Read GTFS files
        print("\nLoading GTFS files...")
        file_load_start = time.time()
        self.stops = pd.read_csv(os.path.join(current_dir, 'stops.txt'))
        self.routes = pd.read_csv(os.path.join(current_dir, 'routes.txt'))
        self.trips = pd.read_csv(os.path.join(current_dir, 'trips.txt'))
        self.stop_times = pd.read_csv(os.path.join(current_dir, 'stop_times.txt'))
        print(f"File loading took {time.time() - file_load_start:.2f} seconds")
        
        # Build route network
        print("\nBuilding network...")
        network_start = time.time()
        self._build_network()
        print(f"Network building took {time.time() - network_start:.2f} seconds")
        
        print(f"\nTotal initialization took {time.time() - start_time:.2f} seconds")
        print(f"Network size: {len(self.stop_routes)} stops and {len(self.route_stops)} routes")
    
    def _build_network(self):
        """Build a network of which stops are connected by which routes"""
        merge_start = time.time()
        
        # Merge stop_times with trips to get route information
        stop_route_data = self.stop_times.merge(
            self.trips[['trip_id', 'route_id']], 
            on='trip_id'
        )[['stop_id', 'route_id']]
        
        print(f"  Merging dataframes took {time.time() - merge_start:.2f} seconds")
        
        # Build mappings using groupby
        mapping_start = time.time()
        self.stop_routes = {
            stop_id: set(group['route_id']) 
            for stop_id, group in stop_route_data.groupby('stop_id')
        }
        self.route_stops = {
            route_id: set(group['stop_id']) 
            for route_id, group in stop_route_data.groupby('route_id')
        }
        print(f"  Building mappings took {time.time() - mapping_start:.2f} seconds")
    
    def find_nearby_stops(self, location_name, limit=3):
        """Find stops that match the given name"""
        matches = self.stops[self.stops['stop_name'].str.contains(location_name, case=False, na=False)]
        return matches.head(limit)
    
    def find_route(self, start_location, end_location, visualize=True):
        """Find a route between two locations with minimum transfers"""
        # Find matching stops
        start_stops = self.find_nearby_stops(start_location)
        end_stops = self.find_nearby_stops(end_location)
        
        if start_stops.empty or end_stops.empty:
            return "Could not find stops matching the provided locations"
        
        end_stop_ids = set(end_stops['stop_id'])
        
        # Try finding a route from each start stop
        best_route = None
        best_transfers = float('inf')
        
        for _, start_stop in start_stops.iterrows():
            route = self._bfs_route(start_stop['stop_id'], end_stop_ids)
            if route and len(route) - 1 < best_transfers:
                best_route = route
                best_transfers = len(route) - 1
        
        if best_route is None:
            return "No route found between the specified locations"
        
        route_description = self._format_route(best_route)
        
        if visualize:
            map_file = self._visualize_route(best_route)
            print(f"\nMap saved to: {map_file}")
            webbrowser.open('file://' + os.path.abspath(map_file))
        
        return route_description
    
    def _bfs_route(self, start_stop_id, end_stop_ids):
        """Use BFS to find route with minimum transfers"""
        queue = deque([(start_stop_id, [start_stop_id], set())])
        visited = {start_stop_id}
        
        while queue:
            current_stop, path, current_routes = queue.popleft()
            
            if current_stop in end_stop_ids:
                return path
            
            routes = self.stop_routes[current_stop]
            
            for route_id in routes:
                for next_stop in self.route_stops[route_id]:
                    if next_stop not in visited:
                        visited.add(next_stop)
                        new_routes = current_routes | {route_id}
                        queue.append((next_stop, path + [next_stop], new_routes))
        
        return None

    def _get_sequential_stops(self, route_id, start_stop_id, end_stop_id):
            """Get sequential stops between start and end stop for a given route"""
            # Get just one trip in each direction for this route
            sample_trips = (
                self.trips[self.trips['route_id'] == route_id]
                .groupby('direction_id', as_index=False)
                .first()
            )
            
            for _, trip in sample_trips.iterrows():
                # Get stops for this trip
                trip_stops = (
                    self.stop_times[self.stop_times['trip_id'] == trip['trip_id']]
                    .sort_values('stop_sequence')
                )
                
                stop_ids = trip_stops['stop_id'].tolist()
                try:
                    start_idx = stop_ids.index(start_stop_id)
                    end_idx = stop_ids.index(end_stop_id)
                    
                    # Handle reverse direction if needed
                    if start_idx > end_idx:
                        return trip_stops.iloc[end_idx:start_idx + 1][::-1]
                    return trip_stops.iloc[start_idx:end_idx + 1]
                except ValueError:
                    continue  # Try next direction
            
            # Fallback: return just the two stops
            return pd.DataFrame({
                'stop_id': [start_stop_id, end_stop_id],
                'stop_sequence': [1, 2]
            })
        
    def _visualize_route(self, stop_ids):
        """Create an interactive map showing the route"""
        first_stop = self.stops[self.stops['stop_id'] == stop_ids[0]].iloc[0]
        m = folium.Map(location=[first_stop['stop_lat'], first_stop['stop_lon']], zoom_start=13)
        
        route_segments = []
        current_segment = []
        current_route_id = None
        
        # Process stops in sequence
        for i in range(len(stop_ids) - 1):
            current_stop_id = stop_ids[i]
            next_stop_id = stop_ids[i + 1]
            
            # Find the route connecting these stops
            common_routes = self.stop_routes[current_stop_id] & self.stop_routes[next_stop_id]
            if common_routes:
                route_id = list(common_routes)[0]
                
                # If this is a new route, start a new segment
                if route_id != current_route_id:
                    if current_segment:
                        route_segments.append((current_route_id, current_segment))
                        current_segment = []
                    current_route_id = route_id
                
                # Get all stops between current and next stop, including intermediates
                sequential_stops = self._get_sequential_stops(route_id, current_stop_id, next_stop_id)
                
                # Add ALL stops in the sequence to build the complete path
                if current_segment and sequential_stops['stop_id'].iloc[0] != current_stop_id:
                    # If we're continuing a segment, skip the first stop to avoid duplicates
                    sequential_stops = sequential_stops.iloc[1:]
                
                for _, stop_row in sequential_stops.iterrows():
                    stop_info = self.stops[self.stops['stop_id'] == stop_row['stop_id']].iloc[0]
                    coords = [stop_info['stop_lat'], stop_info['stop_lon']]
                    if not current_segment or current_segment[-1] != coords:
                        current_segment.append(coords)
        
        # Add the last segment
        if current_segment:
            route_segments.append((current_route_id, current_segment))
        
        # Draw route segments with different colors for different tram lines
        colors = ['blue', 'red', 'green', 'purple', 'orange', 'darkblue']
        for idx, (route_id, segment) in enumerate(route_segments):
            route_name = self.routes[self.routes['route_id'] == route_id]['route_short_name'].iloc[0]
            color = colors[idx % len(colors)]
            
            # Draw the actual path through all stops
            folium.PolyLine(
                locations=segment,
                weight=4,
                color=color,
                popup=f'Tram {route_name}',
                opacity=0.8
            ).add_to(m)
        
        # Add markers only for the main stops (transfers, start, and end)
        for i, stop_id in enumerate(stop_ids):
            stop = self.stops[self.stops['stop_id'] == stop_id].iloc[0]
            
            # Determine marker type
            if i == 0:
                icon_color = 'green'
                popup_prefix = 'Start: '
            elif i == len(stop_ids) - 1:
                icon_color = 'red'
                popup_prefix = 'End: '
            else:
                icon_color = 'blue'
                popup_prefix = 'Transfer: '
            
            folium.Marker(
                [stop['stop_lat'], stop['stop_lon']],
                popup=popup_prefix + stop['stop_name'],
                icon=folium.Icon(color=icon_color)
            ).add_to(m)
        
        # Save map to file
        timestamp = datetime.now().strftime("%Y-%m-%d_%H-%M")
        output_dir = os.path.join('..', 'mpt_data', 'maps')
        os.makedirs(output_dir, exist_ok=True)
        map_file = os.path.join(output_dir, f'tram_route_{timestamp}.html')
        m.save(map_file)
        
        return map_file
    
    def _format_route(self, stop_ids):
        """Format the route into human-readable instructions"""
        route_description = []
        current_route = None
        
        for i in range(len(stop_ids) - 1):
            current_stop = stop_ids[i]
            next_stop = stop_ids[i + 1]
            
            current_routes = self.stop_routes[current_stop] & self.stop_routes[next_stop]
            if current_routes:
                route_id = list(current_routes)[0]
                route_name = self.routes[self.routes['route_id'] == route_id]['route_short_name'].iloc[0]
                
                if route_name != current_route:
                    current_stop_name = self.stops[self.stops['stop_id'] == current_stop]['stop_name'].iloc[0]
                    if current_route:
                        route_description.append(f"Transfer at {current_stop_name}")
                    route_description.append(f"Take tram {route_name} from {current_stop_name}")
                    current_route = route_name
        
        final_stop = self.stops[self.stops['stop_id'] == stop_ids[-1]]['stop_name'].iloc[0]
        route_description.append(f"Arrive at {final_stop}")
        
        return "\n".join(route_description)

# Example usage:
if __name__ == "__main__":
    router = SimpleTramRouter()
    route = router.find_route("Fitzroy", "Prahran")
    print("\nRoute instructions:")
    print(route)
```

```python
import pandas as pd
import os
from collections import defaultdict, deque
import time
import folium
from datetime import datetime
import webbrowser

class SimpleTramRouter:
    def __init__(self):
        start_time = time.time()
        print("Starting router initialization...")
        
        current_dir = os.path.join('..', 'mpt_data', 'tram')
        
        # Read GTFS files
        print("\nLoading GTFS files...")
        file_load_start = time.time()
        self.stops = pd.read_csv(os.path.join(current_dir, 'stops.txt'))
        self.routes = pd.read_csv(os.path.join(current_dir, 'routes.txt'))
        self.trips = pd.read_csv(os.path.join(current_dir, 'trips.txt'))
        self.stop_times = pd.read_csv(os.path.join(current_dir, 'stop_times.txt'))
        self.shapes = pd.read_csv(os.path.join(current_dir, 'shapes.txt'))
        print(f"File loading took {time.time() - file_load_start:.2f} seconds")
        
        # Build route network and shape cache
        print("\nBuilding network...")
        network_start = time.time()
        self._build_network()
        self._build_shape_cache()
        print(f"Network building took {time.time() - network_start:.2f} seconds")
        
        print(f"\nTotal initialization took {time.time() - start_time:.2f} seconds")
        print(f"Network size: {len(self.stop_routes)} stops and {len(self.route_stops)} routes")

    def _build_shape_cache(self):
        """Build a cache of shape points for each route"""
        print("  Building shape cache...")
        shape_start = time.time()
        
        # Get shape_id for each route using trips table
        route_shapes = {}
        for _, trip in self.trips.iterrows():
            route_shapes[trip['route_id']] = trip['shape_id']
        
        # Create cache of shape points
        self.route_shapes = {}
        for route_id, shape_id in route_shapes.items():
            shape_points = (
                self.shapes[self.shapes['shape_id'] == shape_id]
                .sort_values('shape_pt_sequence')[['shape_pt_lat', 'shape_pt_lon']]
                .values.tolist()
            )
            self.route_shapes[route_id] = shape_points
        
        print(f"  Shape cache built in {time.time() - shape_start:.2f} seconds")
    
    def _get_route_segment(self, route_id, start_stop_id, end_stop_id):
        """Get the shape points for a segment of a route between two stops"""
        if route_id not in self.route_shapes:
            return []
            
        # Get stop coordinates
        start_stop = self.stops[self.stops['stop_id'] == start_stop_id].iloc[0]
        end_stop = self.stops[self.stops['stop_id'] == end_stop_id].iloc[0]
        start_coord = [start_stop['stop_lat'], start_stop['stop_lon']]
        end_coord = [end_stop['stop_lat'], end_stop['stop_lon']]
        
        # Get full shape for the route
        shape_points = self.route_shapes[route_id]
        
        # Find closest points on shape to our stops
        start_idx = min(range(len(shape_points)), 
                       key=lambda i: ((shape_points[i][0] - start_coord[0])**2 + 
                                    (shape_points[i][1] - start_coord[1])**2))
        end_idx = min(range(len(shape_points)), 
                     key=lambda i: ((shape_points[i][0] - end_coord[0])**2 + 
                                  (shape_points[i][1] - end_coord[1])**2))
        
        # Return points in correct order
        if start_idx <= end_idx:
            return shape_points[start_idx:end_idx + 1]
        else:
            return shape_points[end_idx:start_idx + 1][::-1]

    def _visualize_route(self, stop_ids):
        """Create an interactive map showing the route"""
        first_stop = self.stops[self.stops['stop_id'] == stop_ids[0]].iloc[0]
        m = folium.Map(location=[first_stop['stop_lat'], first_stop['stop_lon']], zoom_start=13)
        
        route_segments = []
        current_segment = []
        current_route_id = None
        
        # Process stops in sequence
        for i in range(len(stop_ids) - 1):
            current_stop_id = stop_ids[i]
            next_stop_id = stop_ids[i + 1]
            
            # Find the route connecting these stops
            common_routes = self.stop_routes[current_stop_id] & self.stop_routes[next_stop_id]
            if common_routes:
                route_id = list(common_routes)[0]
                
                # If this is a new route, start a new segment
                if route_id != current_route_id:
                    if current_segment:
                        route_segments.append((current_route_id, current_segment))
                        current_segment = []
                    current_route_id = route_id
                
                # Get the shape points for this segment
                segment_points = self._get_route_segment(route_id, current_stop_id, next_stop_id)
                
                # Add to current segment
                if not current_segment:
                    current_segment.extend(segment_points)
                else:
                    current_segment.extend(segment_points[1:])  # Avoid duplicating connecting point
        
        # Add the last segment
        if current_segment:
            route_segments.append((current_route_id, current_segment))
        
        # Draw route segments with different colors for different tram lines
        colors = {
            'blue': '#0066CC',
            'red': '#CC0000',
            'green': '#009933',
            'purple': '#660099',
            'orange': '#FF6600',
        }
        
        for idx, (route_id, segment) in enumerate(route_segments):
            route_name = self.routes[self.routes['route_id'] == route_id]['route_short_name'].iloc[0]
            color = list(colors.values())[idx % len(colors)]
            
            # Draw the route line using shape points
            folium.PolyLine(
                locations=segment,
                weight=4,
                color=color,
                popup=f'Tram {route_name}',
                opacity=0.8
            ).add_to(m)
            
            # Add route number label at midpoint
            if len(segment) >= 2:
                mid_idx = len(segment) // 2
                mid_point = segment[mid_idx]
                folium.DivIcon(
                    html=f'<div style="background-color: {color}; color: white; padding: 3px 6px; border-radius: 3px; font-weight: bold;">Tram {route_name}</div>',
                    icon_size=(50, 20),
                    icon_anchor=(25, 10)
                ).add_to(folium.Marker(mid_point).add_to(m))
        
        # Add markers for main stops
        for i, stop_id in enumerate(stop_ids):
            stop = self.stops[self.stops['stop_id'] == stop_id].iloc[0]
            
            if i == 0:
                icon_color = 'green'
                prefix = 'Start'
            elif i == len(stop_ids) - 1:
                icon_color = 'red'
                prefix = 'End'
            else:
                icon_color = 'blue'
                prefix = 'Transfer'
            
            folium.Marker(
                [stop['stop_lat'], stop['stop_lon']],
                popup=f"{prefix}: {stop['stop_name']}",
                icon=folium.Icon(color=icon_color, icon='info-sign')
            ).add_to(m)
        
        # Save map to file
        timestamp = datetime.now().strftime("%Y-%m-%d_%H-%M")
        output_dir = os.path.join('..', 'mpt_data', 'maps')
        os.makedirs(output_dir, exist_ok=True)
        map_file = os.path.join(output_dir, f'tram_route_{timestamp}.html')
        m.save(map_file)
        
        return map_file

# Example usage:
if __name__ == "__main__":
    router = SimpleTramRouter()
    route = router.find_route("Fitzroy", "Prahran")
    print("\nRoute instructions:")
    print(route)
```

```python
import pandas as pd
import os
from collections import defaultdict, deque
import time
import folium
from datetime import datetime
import webbrowser

class SimpleTramRouter:
    def __init__(self):
        start_time = time.time()
        print("Starting router initialization")
        
        current_dir = os.path.join('..', 'mpt_data', 'tram')
        
        #Read GTFS files
        print("\nLoading GTFS files")
        file_load_start = time.time()
        self.stops = pd.read_csv(os.path.join(current_dir, 'stops.txt'))
        self.routes = pd.read_csv(os.path.join(current_dir, 'routes.txt'))
        self.trips = pd.read_csv(os.path.join(current_dir, 'trips.txt'))
        self.stop_times = pd.read_csv(os.path.join(current_dir, 'stop_times.txt'))
        self.shapes = pd.read_csv(os.path.join(current_dir, 'shapes.txt'))
        print(f"File loading took {time.time() - file_load_start:.2f} seconds")
        
        #Build route network and shape cache
        print("\nBuilding network")
        network_start = time.time()
        self._build_network()
        self._build_shape_cache()
        print(f"Network building took {time.time() - network_start:.2f} seconds")
        
        print(f"\nTotal initialization took {time.time() - start_time:.2f} seconds")
        print(f"Network size: {len(self.stop_routes)} stops and {len(self.route_stops)} routes")

    def _build_network(self):
        merge_start = time.time()
        
        #Merge stop_times with trips to get route information
        stop_route_data = self.stop_times.merge(
            self.trips[['trip_id', 'route_id']], 
            on='trip_id'
        )[['stop_id', 'route_id']]
        
        print(f"  Merging dataframes took {time.time() - merge_start:.2f} seconds")
        
        #Build mappings using groupby
        mapping_start = time.time()
        self.stop_routes = {
            stop_id: set(group['route_id']) 
            for stop_id, group in stop_route_data.groupby('stop_id')
        }
        self.route_stops = {
            route_id: set(group['stop_id']) 
            for route_id, group in stop_route_data.groupby('route_id')
        }
        print(f"  Building mappings took {time.time() - mapping_start:.2f} seconds")

    def _build_shape_cache(self):
        print("  Building shape cache")
        shape_start = time.time()
        
        #Get shape_id for each route using trips table
        route_shapes = {}
        for _, trip in self.trips.iterrows():
            route_shapes[trip['route_id']] = trip['shape_id']
        
        #Create cache of shape points
        self.route_shapes = {}
        for route_id, shape_id in route_shapes.items():
            shape_points = (
                self.shapes[self.shapes['shape_id'] == shape_id]
                .sort_values('shape_pt_sequence')[['shape_pt_lat', 'shape_pt_lon']]
                .values.tolist()
            )
            self.route_shapes[route_id] = shape_points
        
        print(f"  Shape cache built in {time.time() - shape_start:.2f} seconds")
    
    def find_nearby_stops(self, location_name, limit=3):
        """Find stops that match the given name"""
        matches = self.stops[self.stops['stop_name'].str.contains(location_name, case=False, na=False)]
        return matches.head(limit)
    
    def find_route(self, start_location, end_location, visualize=True):
        """Find a route between two locations with minimum transfers"""
        start_stops = self.find_nearby_stops(start_location)
        end_stops = self.find_nearby_stops(end_location)
        
        if start_stops.empty or end_stops.empty:
            return "Could not find stops matching the provided locations"
        
        end_stop_ids = set(end_stops['stop_id'])
        
        #Try finding a route from each start stop
        best_route = None
        best_transfers = float('inf')
        
        for _, start_stop in start_stops.iterrows():
            route = self._bfs_route(start_stop['stop_id'], end_stop_ids)
            if route and len(route) - 1 < best_transfers:
                best_route = route
                best_transfers = len(route) - 1
        
        if best_route is None:
            return "No route found between the specified locations"
        
        route_description = self._format_route(best_route)
        
        if visualize:
            map_file = self._visualize_route(best_route)
            print(f"\nMap saved to: {map_file}")
            webbrowser.open('file://' + os.path.abspath(map_file))
        
        return route_description
    
    def _bfs_route(self, start_stop_id, end_stop_ids):

        queue = deque([(start_stop_id, [start_stop_id], set())])
        visited = {start_stop_id}
        
        while queue:
            current_stop, path, current_routes = queue.popleft()
            
            if current_stop in end_stop_ids:
                return path
            
            routes = self.stop_routes[current_stop]
            
            for route_id in routes:
                for next_stop in self.route_stops[route_id]:
                    if next_stop not in visited:
                        visited.add(next_stop)
                        new_routes = current_routes | {route_id}
                        queue.append((next_stop, path + [next_stop], new_routes))
        
        return None

    def _get_route_segment(self, route_id, start_stop_id, end_stop_id):
        """Get the shape points for a segment of a route between two stops"""
        if route_id not in self.route_shapes:
            return []
            
        #Get stop coords
        start_stop = self.stops[self.stops['stop_id'] == start_stop_id].iloc[0]
        end_stop = self.stops[self.stops['stop_id'] == end_stop_id].iloc[0]
        start_coord = [start_stop['stop_lat'], start_stop['stop_lon']]
        end_coord = [end_stop['stop_lat'], end_stop['stop_lon']]
        
        #Get full shape for the route
        shape_points = self.route_shapes[route_id]
        
        #Find closest points on shape to our stops
        start_idx = min(range(len(shape_points)), 
                       key=lambda i: ((shape_points[i][0] - start_coord[0])**2 + 
                                    (shape_points[i][1] - start_coord[1])**2))
        end_idx = min(range(len(shape_points)), 
                     key=lambda i: ((shape_points[i][0] - end_coord[0])**2 + 
                                  (shape_points[i][1] - end_coord[1])**2))
        
        # Return points in correct order
        if start_idx <= end_idx:
            return shape_points[start_idx:end_idx + 1]
        else:
            return shape_points[end_idx:start_idx + 1][::-1]

    def _visualize_route(self, stop_ids):
        """Creating map showing the route"""
        first_stop = self.stops[self.stops['stop_id'] == stop_ids[0]].iloc[0]
        m = folium.Map(location=[first_stop['stop_lat'], first_stop['stop_lon']], zoom_start=13)
        
        route_segments = []
        current_segment = []
        current_route_id = None
        
        # Process stops in sequence
        for i in range(len(stop_ids) - 1):
            current_stop_id = stop_ids[i]
            next_stop_id = stop_ids[i + 1]
            
            # Find the route connecting these stops
            common_routes = self.stop_routes[current_stop_id] & self.stop_routes[next_stop_id]
            if common_routes:
                route_id = list(common_routes)[0]
                
                #If this is a new route, start a new segment
                if route_id != current_route_id:
                    if current_segment:
                        route_segments.append((current_route_id, current_segment))
                        current_segment = []
                    current_route_id = route_id
                
                #Get the shape points for this segment
                segment_points = self._get_route_segment(route_id, current_stop_id, next_stop_id)
                
                # Add to current segment
                if not current_segment:
                    current_segment.extend(segment_points)
                else:
                    current_segment.extend(segment_points[1:])  # Avoid duplicating connecting point
        
        #Add the last segment
        if current_segment:
            route_segments.append((current_route_id, current_segment))
        
        # Draw route segments with different colors for different tram lines
        colors = {
            'blue': '#0066CC',
            'red': '#CC0000',
            'green': '#009933',
            'purple': '#660099',
            'orange': '#FF6600',
        }
        
        for idx, (route_id, segment) in enumerate(route_segments):
            route_name = self.routes[self.routes['route_id'] == route_id]['route_short_name'].iloc[0]
            color = list(colors.values())[idx % len(colors)]
            
            #Draw the route line using shape points
            folium.PolyLine(
                locations=segment,
                weight=4,
                color=color,
                popup=f'Tram {route_name}',
                opacity=0.8
            ).add_to(m)
            
            #Add route number label at midpoint
            if len(segment) >= 2:
                mid_idx = len(segment) // 2
                mid_point = segment[mid_idx]
                folium.DivIcon(
                    html=f'<div style="background-color: {color}; color: white; padding: 3px 6px; border-radius: 3px; font-weight: bold;">Tram {route_name}</div>',
                    icon_size=(50, 20),
                    icon_anchor=(25, 10)
                ).add_to(folium.Marker(mid_point).add_to(m))
        
        #Add markers for main stops
        for i, stop_id in enumerate(stop_ids):
            stop = self.stops[self.stops['stop_id'] == stop_id].iloc[0]
            
            if i == 0:
                icon_color = 'green'
                prefix = 'Start'
            elif i == len(stop_ids) - 1:
                icon_color = 'red'
                prefix = 'End'
            else:
                icon_color = 'blue'
                prefix = 'Transfer'
            
            folium.Marker(
                [stop['stop_lat'], stop['stop_lon']],
                popup=f"{prefix}: {stop['stop_name']}",
                icon=folium.Icon(color=icon_color, icon='info-sign')
            ).add_to(m)
        
        #Save map to file
        timestamp = datetime.now().strftime("%Y-%m-%d_%H-%M")
        output_dir = os.path.join('..', 'mpt_data', 'maps')
        os.makedirs(output_dir, exist_ok=True)
        map_file = os.path.join(output_dir, f'tram_route_{timestamp}.html')
        m.save(map_file)
        
        return map_file
    
    def _format_route(self, stop_ids):
        """Format the route into human-readable instructions"""
        route_description = []
        current_route = None
        
        for i in range(len(stop_ids) - 1):
            current_stop = stop_ids[i]
            next_stop = stop_ids[i + 1]
            
            current_routes = self.stop_routes[current_stop] & self.stop_routes[next_stop]
            if current_routes:
                route_id = list(current_routes)[0]
                route_name = self.routes[self.routes['route_id'] == route_id]['route_short_name'].iloc[0]
                
                if route_name != current_route:
                    current_stop_name = self.stops[self.stops['stop_id'] == current_stop]['stop_name'].iloc[0]
                    if current_route:
                        route_description.append(f"Transfer at {current_stop_name}")
                    route_description.append(f"Take tram {route_name} from {current_stop_name}")
                    current_route = route_name
        
        final_stop = self.stops[self.stops['stop_id'] == stop_ids[-1]]['stop_name'].iloc[0]
        route_description.append(f"Arrive at {final_stop}")
        
        return "\n".join(route_description)

#Test
#if __name__ == "__main__":
#    router = SimpleTramRouter()
#    route = router.find_route("Essendon", "Prahran")
#    print("\nRoute instructions:")
#    print(route)


if __name__ == "__main__":
  
#    print("Welcome to the Simple Tram Router!")
    print("Please enter your start and end locations to find a tram route.\n")
    
    # Take user input for start and end locations
    start_location = input("Enter the starting location: ").strip()
    end_location = input("Enter the destination location: ").strip()

    # Initialize the router and find the route
    router = SimpleTramRouter()
    route = router.find_route(start_location, end_location)
    
    print("\nRoute instructions:")
    print(route)
```

```python
import pandas as pd
import os
from collections import deque
import time
import folium
from datetime import datetime
import webbrowser


class SimpleTramRouter:
    @staticmethod
    def normalise_gtfs_data(stops_df: pd.DataFrame, stop_times_df: pd.DataFrame) -> None:
        """
        Normalize the stop names and ensure the stop_times DataFrame is indexed correctly.
        """
        # Normalize stop names and IDs
        stops_df['stop_name'] = stops_df['stop_name'].astype(str).str.strip()
        stops_df['stop_id'] = stops_df['stop_id'].astype(str).str.strip()
        stops_df['normalized_stop_name'] = stops_df['stop_name'].str.lower()

        # Normalize stop_times IDs and set index
        stop_times_df['stop_id'] = stop_times_df['stop_id'].astype(str).str.strip()
        expected_columns = ['stop_id', 'trip_id', 'arrival_time', 'departure_time']

        if all(col in stop_times_df.columns for col in expected_columns):
            try:
                stop_times_df.set_index(['stop_id', 'trip_id'], inplace=True)
            except KeyError as e:
                print(f"Error setting index: {e}")
        else:
            print("Expected columns are not present in the DataFrame.")
            print(f"Available columns: {stop_times_df.columns}")

    def __init__(self):
        start_time = time.time()
        print("Starting router initialization")
        
        current_dir = os.path.join('..', 'mpt_data', 'tram')
        
        # Read GTFS files
        print("\nLoading GTFS files")
        file_load_start = time.time()
        self.stops = pd.read_csv(os.path.join(current_dir, 'stops.txt'))
        self.routes = pd.read_csv(os.path.join(current_dir, 'routes.txt'))
        self.trips = pd.read_csv(os.path.join(current_dir, 'trips.txt'))
        self.stop_times = pd.read_csv(os.path.join(current_dir, 'stop_times.txt'))
        self.shapes = pd.read_csv(os.path.join(current_dir, 'shapes.txt'))
        print(f"File loading took {time.time() - file_load_start:.2f} seconds")

        # Normalize GTFS data
        print("\nNormalizing GTFS data")
        self.normalise_gtfs_data(self.stops, self.stop_times)

        # Build route network and shape cache
        print("\nBuilding network")
        network_start = time.time()
        self._build_network()
        self._build_shape_cache()
        print(f"Network building took {time.time() - network_start:.2f} seconds")
        
        print(f"\nTotal initialization took {time.time() - start_time:.2f} seconds")
        print(f"Network size: {len(self.stop_routes)} stops and {len(self.route_stops)} routes")
        print("Routes in system:", self.routes['route_id'].tolist())


    def _build_network(self):
        merge_start = time.time()
        
        # Merge stop_times with trips to get route information
        stop_route_data = self.stop_times.reset_index().merge(
            self.trips[['trip_id', 'route_id']], 
            on='trip_id'
        )[['stop_id', 'route_id']]
        
        print(f"  Merging dataframes took {time.time() - merge_start:.2f} seconds")
        
        # Build mappings using groupby
        mapping_start = time.time()
        self.stop_routes = {
            stop_id: set(group['route_id']) 
            for stop_id, group in stop_route_data.groupby('stop_id')
        }
        self.route_stops = {
            route_id: set(group['stop_id']) 
            for route_id, group in stop_route_data.groupby('route_id')
        }
        print(f"  Building mappings took {time.time() - mapping_start:.2f} seconds")

    def _build_shape_cache(self):
        print("  Building shape cache")
        shape_start = time.time()
        
        # Get shape_id for each route using trips table
        route_shapes = {}
        for _, trip in self.trips.iterrows():
            route_shapes[trip['route_id']] = trip['shape_id']
        
        # Create cache of shape points
        self.route_shapes = {}
        for route_id, shape_id in route_shapes.items():
            shape_points = (
                self.shapes[self.shapes['shape_id'] == shape_id]
                .sort_values('shape_pt_sequence')[['shape_pt_lat', 'shape_pt_lon']]
                .values.tolist()
            )
            self.route_shapes[route_id] = shape_points
        
        print(f"  Shape cache built in {time.time() - shape_start:.2f} seconds")

    def _bfs_route(self, start_stop_id, end_stop_ids):
        """Perform BFS to find the shortest route from start_stop_id to any of the end_stop_ids."""
        queue = deque([(start_stop_id, [start_stop_id], set())])  # (current_stop, path, current_routes)
        visited = {start_stop_id}  # Keep track of visited stops

        while queue:
            current_stop, path, current_routes = queue.popleft()

            # Check if we've reached one of the destination stops
            if current_stop in end_stop_ids:
                return path

            # Get all routes passing through the current stop
            routes = self.stop_routes[current_stop]

            for route_id in routes:
                # Traverse all stops on this route
                for next_stop in self.route_stops[route_id]:
                    if next_stop not in visited:
                        visited.add(next_stop)
                        new_routes = current_routes | {route_id}
                        queue.append((next_stop, path + [next_stop], new_routes))

        return None  # Return None if no route is found

    def find_nearby_stops(self, location_name, limit=3):
        """Find stops that match the given name"""
        location_name = location_name.lower().strip()
        matches = self.stops[self.stops['normalized_stop_name'].str.contains(location_name, case=False, na=False)]
        return matches.head(limit)

    def find_route(self, start_location, end_location, visualize=True):
        """Find a route between two locations with minimum transfers"""
        start_stops = self.find_nearby_stops(start_location)
        end_stops = self.find_nearby_stops(end_location)
        
        if start_stops.empty or end_stops.empty:
            return "Could not find stops matching the provided locations"
        
        end_stop_ids = set(end_stops['stop_id'])
        
        # Try finding a route from each start stop
        best_route = None
        best_transfers = float('inf')
        
        for _, start_stop in start_stops.iterrows():
            route = self._bfs_route(start_stop['stop_id'], end_stop_ids)
            if route and len(route) - 1 < best_transfers:
                best_route = route
                best_transfers = len(route) - 1
        
        if best_route is None:
            return "No route found between the specified locations"
        
        route_description = self._format_route(best_route)
        
        if visualize:
            map_file = self._visualize_route(best_route)
            print(f"\nMap saved to: {map_file}")
            webbrowser.open('file://' + os.path.abspath(map_file))
        
        return route_description

    # (Other helper methods remain unchanged)


    def _format_route(self, stop_ids):
        """Format the route into human-readable instructions"""
        route_description = []
        current_route = None
        
        for i in range(len(stop_ids) - 1):
            current_stop = stop_ids[i]
            next_stop = stop_ids[i + 1]
            
            current_routes = self.stop_routes[current_stop] & self.stop_routes[next_stop]
            if current_routes:
                route_id = list(current_routes)[0]
                route_name = self.routes[self.routes['route_id'] == route_id]['route_short_name'].iloc[0]
                
                if route_name != current_route:
                    current_stop_name = self.stops[self.stops['stop_id'] == current_stop]['stop_name'].iloc[0]
                    if current_route:
                        route_description.append(f"Transfer at {current_stop_name}")
                    route_description.append(f"Take tram {route_name} from {current_stop_name}")
                    current_route = route_name
        
        final_stop = self.stops[self.stops['stop_id'] == stop_ids[-1]]['stop_name'].iloc[0]
        route_description.append(f"Arrive at {final_stop}")
        
        return "\n".join(route_description)

    def _visualize_route(self, stop_ids):
        """Creating map showing the route"""
        first_stop = self.stops[self.stops['stop_id'] == stop_ids[0]].iloc[0]
        m = folium.Map(location=[first_stop['stop_lat'], first_stop['stop_lon']], zoom_start=13)
        
        route_segments = []
        current_segment = []
        current_route_id = None
        
        # Process stops in sequence
        for i in range(len(stop_ids) - 1):
            current_stop_id = stop_ids[i]
            next_stop_id = stop_ids[i + 1]
            
            # Find the route connecting these stops
            common_routes = self.stop_routes[current_stop_id] & self.stop_routes[next_stop_id]
            if common_routes:
                route_id = list(common_routes)[0]
                
                #If this is a new route, start a new segment
                if route_id != current_route_id:
                    if current_segment:
                        route_segments.append((current_route_id, current_segment))
                        current_segment = []
                    current_route_id = route_id
                
                #Get the shape points for this segment
                segment_points = self._get_route_segment(route_id, current_stop_id, next_stop_id)
                
                # Add to current segment
                if not current_segment:
                    current_segment.extend(segment_points)
                else:
                    current_segment.extend(segment_points[1:])  # Avoid duplicating connecting point
        
        #Add the last segment
        if current_segment:
            route_segments.append((current_route_id, current_segment))
        
        # Draw route segments with different colors for different tram lines
        colors = {
            'blue': '#0066CC',
            'red': '#CC0000',
            'green': '#009933',
            'purple': '#660099',
            'orange': '#FF6600',
        }
        
        for idx, (route_id, segment) in enumerate(route_segments):
            route_name = self.routes[self.routes['route_id'] == route_id]['route_short_name'].iloc[0]
            color = list(colors.values())[idx % len(colors)]
            
            #Draw the route line using shape points
            folium.PolyLine(
                locations=segment,
                weight=4,
                color=color,
                popup=f'Tram {route_name}',
                opacity=0.8
            ).add_to(m)
            
            #Add route number label at midpoint
            if len(segment) >= 2:
                mid_idx = len(segment) // 2
                mid_point = segment[mid_idx]
                folium.DivIcon(
                    html=f'<div style="background-color: {color}; color: white; padding: 3px 6px; border-radius: 3px; font-weight: bold;">Tram {route_name}</div>',
                    icon_size=(50, 20),
                    icon_anchor=(25, 10)
                ).add_to(folium.Marker(mid_point).add_to(m))
        
        #Add markers for main stops
        for i, stop_id in enumerate(stop_ids):
            stop = self.stops[self.stops['stop_id'] == stop_id].iloc[0]
            
            if i == 0:
                icon_color = 'green'
                prefix = 'Start'
            elif i == len(stop_ids) - 1:
                icon_color = 'red'
                prefix = 'End'
            else:
                icon_color = 'blue'
                prefix = 'Transfer'
            
            folium.Marker(
                [stop['stop_lat'], stop['stop_lon']],
                popup=f"{prefix}: {stop['stop_name']}",
                icon=folium.Icon(color=icon_color, icon='info-sign')
            ).add_to(m)
        
        #Save map to file
        timestamp = datetime.now().strftime("%Y-%m-%d_%H-%M")
        output_dir = os.path.join('..', 'mpt_data', 'maps')
        os.makedirs(output_dir, exist_ok=True)
        map_file = os.path.join(output_dir, f'tram_route_{timestamp}.html')
        m.save(map_file)
        
        return map_file

    def _get_route_segment(self, route_id, start_stop_id, end_stop_id):
        """Get the shape points for a segment of a route between two stops"""
        if route_id not in self.route_shapes:
            return []
            
        #Get stop coords
        start_stop = self.stops[self.stops['stop_id'] == start_stop_id].iloc[0]
        end_stop = self.stops[self.stops['stop_id'] == end_stop_id].iloc[0]
        start_coord = [start_stop['stop_lat'], start_stop['stop_lon']]
        end_coord = [end_stop['stop_lat'], end_stop['stop_lon']]
        
        #Get full shape for the route
        shape_points = self.route_shapes[route_id]
        
        #Find closest points on shape to our stops
        start_idx = min(range(len(shape_points)), 
                       key=lambda i: ((shape_points[i][0] - start_coord[0])**2 + 
                                    (shape_points[i][1] - start_coord[1])**2))
        end_idx = min(range(len(shape_points)), 
                     key=lambda i: ((shape_points[i][0] - end_coord[0])**2 + 
                                  (shape_points[i][1] - end_coord[1])**2))
        
        # Return points in correct order
        if start_idx <= end_idx:
            return shape_points[start_idx:end_idx + 1]
        else:
            return shape_points[end_idx:start_idx + 1][::-1]

# Main script
if __name__ == "__main__":
    # Prompt the user for input
    print("Please enter your start and end locations to find a tram route.\n")
    
    start_location = input("Enter the starting location: ").strip()
    end_location = input("Enter the destination location: ").strip()

    # Initialize the router and find the route
    router = SimpleTramRouter()
    route = router.find_route(start_location, end_location)
    
    print("\nRoute instructions:")
    print(route)
```
