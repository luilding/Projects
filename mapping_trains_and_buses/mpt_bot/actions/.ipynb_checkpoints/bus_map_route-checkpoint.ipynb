```python
import requests
import folium
from geopy.geocoders import Nominatim
from datetime import datetime
import pandas as pd
from scipy.spatial import KDTree
from geopy.distance import geodesic
import os
import webbrowser

def get_project_root():
    """Get the project root directory by going up one level from the current directory"""
    current_dir = os.getcwd()
    return os.path.dirname(current_dir)

def ensure_directory_exists(directory):
    """Create directory if it doesn't exist"""
    if not os.path.exists(directory):
        os.makedirs(directory)

def load_gtfs_data(project_root):
    """Load all required GTFS files"""
    data_dir = os.path.join(project_root, 'mpt_data', 'bus')
    
    stops_df = pd.read_csv(os.path.join(data_dir, 'stops.txt'))
    routes_df = pd.read_csv(os.path.join(data_dir, 'routes.txt'))
    trips_df = pd.read_csv(os.path.join(data_dir, 'trips.txt'))
    stop_times_df = pd.read_csv(os.path.join(data_dir, 'stop_times.txt'))
    shapes_df = pd.read_csv(os.path.join(data_dir, 'shapes.txt')) if os.path.exists(os.path.join(data_dir, 'shapes.txt')) else None
    
    print(f"Loaded {len(stops_df)} stops")
    print(f"Loaded {len(routes_df)} routes")
    print(f"Loaded {len(trips_df)} trips")
    print(f"Loaded {len(stop_times_df)} stop times")
    
    return stops_df, routes_df, trips_df, stop_times_df, shapes_df

def find_direct_route(stop_times_df, trips_df, routes_df, start_stop_id, end_stop_id):
    """Find direct bus routes between two stops"""
    print(f"\nSearching for routes between stop {start_stop_id} and {end_stop_id}")
    
    #Get all trips that contain both stops
    start_trips = set(stop_times_df[stop_times_df['stop_id'] == start_stop_id]['trip_id'])
    end_trips = set(stop_times_df[stop_times_df['stop_id'] == end_stop_id]['trip_id'])
    common_trips = start_trips & end_trips
    
    print(f"Found {len(start_trips)} trips from start stop")
    print(f"Found {len(end_trips)} trips from end stop")
    print(f"Found {len(common_trips)} common trips")
    
    if not common_trips:
        return None
    
    valid_routes = []
    for trip_id in common_trips:
        trip_stops = stop_times_df[stop_times_df['trip_id'] == trip_id].sort_values('stop_sequence')
        
        #Check if start stop comes before end stop in sequence
        start_seq = trip_stops[trip_stops['stop_id'] == start_stop_id]['stop_sequence'].iloc[0]
        end_seq = trip_stops[trip_stops['stop_id'] == end_stop_id]['stop_sequence'].iloc[0]
        
        if start_seq < end_seq:
            route_id = trips_df[trips_df['trip_id'] == trip_id]['route_id'].iloc[0]
            route_info = routes_df[routes_df['route_id'] == route_id].iloc[0]
            
            #Get times
            start_time = trip_stops[trip_stops['stop_id'] == start_stop_id]['departure_time'].iloc[0]
            end_time = trip_stops[trip_stops['stop_id'] == end_stop_id]['arrival_time'].iloc[0]
            
            valid_routes.append({
                'route_name': route_info['route_long_name'],
                'route_number': route_info['route_short_name'],
                'start_time': start_time,
                'end_time': end_time,
                'trip_id': trip_id,
                'route_id': route_id
            })
    
    print(f"Found {len(valid_routes)} valid routes")
    return valid_routes if valid_routes else None

def get_route_shape(shapes_df, route_id, trips_df):
    """Get the shape points for a specific route"""
    if shapes_df is None:
        return None
    
    #Get shape_id from trips for this route
    trip = trips_df[trips_df['route_id'] == route_id]
    if trip.empty:
        return None
    
    shape_id = trip.iloc[0]['shape_id']
    
    #Get shape points
    shape_points = shapes_df[shapes_df['shape_id'] == shape_id].sort_values('shape_pt_sequence')
    if shape_points.empty:
        return None
        
    return shape_points[['shape_pt_lat', 'shape_pt_lon']].values.tolist()

def geocode_address(address):
    geolocator = Nominatim(user_agent="bus_mapping_app1.0")
    melbourne_bbox = [(-38.5267, 144.5937), (-37.5113, 145.5125)]
    
    # Add some common location mappings
    location_mappings = {
        'mcg': 'Melbourne Cricket Ground, Melbourne',
        'melbourne cricket ground': 'Melbourne Cricket Ground, Melbourne'
    }
    
    # Check if we have a mapping for this address
    search_address = location_mappings.get(address.lower(), address + ', Melbourne, Australia')
    
    location = geolocator.geocode(search_address, viewbox=melbourne_bbox, bounded=True)
    
    if location:
        print(f"Found location: {location.address}")
        print(f"Coordinates: {location.latitude}, {location.longitude}")
        return location.latitude, location.longitude
    else:
        print(f"Address '{address}' not found within Melbourne.")
        return None

def find_nearest_stop(lat, lon, kdtree, df):
    distance, index = kdtree.query([lat, lon])
    nearest_stop = df.iloc[index]
    
    stop_coords = (nearest_stop["stop_lat"], nearest_stop["stop_lon"])
    point_coords = (lat, lon)
    distance_meters = geodesic(point_coords, stop_coords).meters
    
    print(f"\nNearest stop: {nearest_stop['stop_name']}")
    print(f"Stop ID: {nearest_stop['stop_id']}")
    print(f"Distance: {distance_meters:.0f} meters")
    
    return nearest_stop, distance_meters, distance

def visualize_route(start_location, start_stop, end_location, end_stop, route_info=None, route_shape=None):
    m = folium.Map(location=[start_location[0], start_location[1]], zoom_start=13)
    
    # Add markers for start and end locations
    folium.Marker(
        [start_location[0], start_location[1]],
        popup="Start",
        icon=folium.Icon(color='green')
    ).add_to(m)
    
    folium.Marker(
        [end_location[0], end_location[1]],
        popup="Destination",
        icon=folium.Icon(color='red')
    ).add_to(m)
    
    # Add markers for bus stops with route information
    start_popup = f"Bus Stop: {start_stop['stop_name']}"
    if route_info:
        start_popup += f"<br>Route: {route_info['route_number']}"
        start_popup += f"<br>Departure: {route_info['start_time']}"
    
    folium.Marker(
        [start_stop["stop_lat"], start_stop["stop_lon"]],
        popup=start_popup,
        icon=folium.Icon(color='blue')
    ).add_to(m)
    
    end_popup = f"Bus Stop: {end_stop['stop_name']}"
    if route_info:
        end_popup += f"<br>Arrival: {route_info['end_time']}"
    
    folium.Marker(
        [end_stop["stop_lat"], end_stop["stop_lon"]],
        popup=end_popup,
        icon=folium.Icon(color='blue')
    ).add_to(m)
    
    # Draw walking lines
    folium.PolyLine(
        locations=[[start_location[0], start_location[1]], 
                  [start_stop["stop_lat"], start_stop["stop_lon"]]],
        weight=2,
        color='green',
        opacity=0.8,
        popup='Walk to bus stop'
    ).add_to(m)
    
    # Draw bus route
    if route_shape:
        folium.PolyLine(
            locations=route_shape,
            weight=3,
            color='blue',
            opacity=0.8,
            popup=f"Bus route {route_info['route_number'] if route_info else ''}"
        ).add_to(m)
    else:
        folium.PolyLine(
            locations=[[start_stop["stop_lat"], start_stop["stop_lon"]], 
                      [end_stop["stop_lat"], end_stop["stop_lon"]]],
            weight=3,
            color='blue',
            opacity=0.8,
            popup=f"Bus route {route_info['route_number'] if route_info else ''}"
        ).add_to(m)
    
    folium.PolyLine(
        locations=[[end_stop["stop_lat"], end_stop["stop_lon"]], 
                  [end_location[0], end_location[1]]],
        weight=2,
        color='red',
        opacity=0.8,
        popup='Walk to destination'
    ).add_to(m)
    
    output_dir = os.path.join(get_project_root(), 'mpt_data')
    ensure_directory_exists(output_dir)
    
    timestamp = datetime.now().strftime("%Y-%m-%d_%H-%M")
    map_file = os.path.join(output_dir, f'bus_directions_{timestamp}.html')
    
    m.save(map_file)
    return map_file

def main():
    print("\nBus Route Planner")
    print("-----------------")
    
    try:
        # Get the project root directory and load all GTFS data
        project_root = get_project_root()
        stops_df, routes_df, trips_df, stop_times_df, shapes_df = load_gtfs_data(project_root)
        
        # Create the k-d tree
        coords = stops_df[["stop_lat", "stop_lon"]].values
        kdtree = KDTree(coords)
        
        # Get user input
        current_location = input("\nEnter your starting location: ")
        destination_input = input("Enter your destination: ")
        
        print("\nProcessing your request...")
        
        # Geocode locations and find nearest stops
        start_coords = geocode_address(current_location)
        if not start_coords:
            raise ValueError("Could not find starting location.")
            
        nearest_start_stop, start_distance, _ = find_nearest_stop(start_coords[0], start_coords[1], kdtree, stops_df)
        
        end_coords = geocode_address(destination_input)
        if not end_coords:
            raise ValueError("Could not find destination location.")
        
        nearest_end_stop, end_distance, _ = find_nearest_stop(end_coords[0], end_coords[1], kdtree, stops_df)
        
        # Find available routes
        routes = find_direct_route(stop_times_df, trips_df, routes_df, 
                                 nearest_start_stop['stop_id'], 
                                 nearest_end_stop['stop_id'])
        
        if routes:
            # Use the first available route for now
            best_route = routes[0]
            
            # Get route shape
            route_shape = get_route_shape(shapes_df, best_route['route_id'], trips_df)
            
            # Generate the map with route information
            map_file = visualize_route(
                start_coords, 
                nearest_start_stop, 
                end_coords, 
                nearest_end_stop,
                best_route,
                route_shape
            )
            
            # Print route information
            print("\nRoute Details:")
            print(f"1. Walk {start_distance:.0f} meters to {nearest_start_stop['stop_name']} bus stop")
            print(f"2. Take bus {best_route['route_number']} ({best_route['route_name']})")
            print(f"   Departure: {best_route['start_time']}")
            print(f"   Arrival: {best_route['end_time']}")
            print(f"3. Walk {end_distance:.0f} meters to your destination")
            
            # Show alternative routes if available
            if len(routes) > 1:
                print("\nAlternative routes available:")
                for route in routes[1:]:
                    print(f"- Bus {route['route_number']} ({route['route_name']})")
                    print(f"  Departure: {route['start_time']}, Arrival: {route['end_time']}")
        else:
            print("\nNo direct bus routes found between these stops.")
            
            # Let's provide more information about why no routes were found
            print("\nDebugging information:")
            print(f"Start stop ID: {nearest_start_stop['stop_id']}")
            print(f"End stop ID: {nearest_end_stop['stop_id']}")
            print("Try different stops or consider multiple bus transfers.")
            
            # Generate basic map without route information
            map_file = visualize_route(
                start_coords, 
                nearest_start_stop, 
                end_coords, 
                nearest_end_stop
            )
        
        print(f"\nMap saved as: {map_file}")
        webbrowser.open('file://' + os.path.abspath(map_file))
        
    except FileNotFoundError as e:
        print(f"Error: {str(e)}")
    except Exception as e:
        print(f"An error occurred: {str(e)}")

if __name__ == "__main__":
    main()
```
